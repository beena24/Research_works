```python
# Creating multi-dimentional arrays into 2D dataframes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Data Distributional Plot
import seaborn as sns
```

```python
# Read the Dataset
boston_df = pd.read_csv("https://raw.githubusercontent.com/ammishra08/MachineLearning/master/Datasets/boston_train.csv", sep = ',')
```

```python
# MEDV denotes median value of house in 1000$
boston_df.head()
```

```python
boston_df.isnull()
```

```python
boston_df.isnull().sum()
```

##### Line Plot

```python
plt.figure(figsize = (18,9))
plt.plot(boston_df['AGE'])
plt.title("Age Line Plot", fontsize = 12)
plt.xlabel("Row Indexes")
plt.ylabel("Age of House")
plt.show()
```

```python
boston_df['AGE'].sort_values().unique()
```

```python
plt.figure(figsize = (11,7))
plt.plot(boston_df['AGE'].sort_values().unique())
plt.title("Age Line Plot", fontsize = 12)
plt.xlabel("Row Indexes")
plt.ylabel("Age of House")
plt.show()
```

##### Bar Plot
    * Bivariate Visulaization - x must be discrete & y variable can be discrete or continious

```python
plt.figure(figsize = (14,8))
plt.bar(x = boston_df['ZN'], height = boston_df['AGE'], width = 1.5, color = 'deeppink')
plt.title("Bar Plot", fontsize = 12)
plt.xlabel("ZN")
plt.ylabel("Max. of AGE")
plt.show()
```

##### Histogram Plot
    * Displays Spread/Distribution of Data, Univariate Analysis & Continious Data

```python
plt.figure(figsize = (12,7))
# Bins - Number of Intervals within range
plt.hist(boston_df['MEDV'], bins = 15, color = 'lightgreen')
plt.title("Histogram Plot", fontsize = 12)
plt.xlabel("MEDV")
plt.ylabel("Frequency in Range")
plt.show()
```

##### Scatter Plot
    * Bivariate Plotting of X & Y

```python
iris_df = pd.read_csv("https://raw.githubusercontent.com/ammishra08/MachineLearning/master/Datasets/iris.csv")
display(iris_df)
```

```python
plt.figure(figsize = (12,7))
plt.scatter(x = iris_df['sepal_length'], y = iris_df['petal_length'], c = 'darkviolet')
plt.title("Scatter Plot", fontsize = 12)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
plt.style.use("seaborn-whitegrid")
plt.figure(figsize = (13,9))
plt.scatter(x = iris_df['sepal_length'], y = iris_df['petal_length'], label = "Petal Length", c = 'deeppink', s = 60, marker = '^')
plt.scatter(x = iris_df['sepal_length'], y = iris_df['petal_width'], label = "Petal Width", c = 'deepskyblue', s = 60, marker = 's')
plt.title("Scatter Plot", fontsize = 12)
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length/Petal Width")
plt.legend()
plt.show()
```

```python
sns.set_style("whitegrid")
plt.figure(figsize = (13,9))
sns.scatterplot(x = 'sepal_length', y = 'petal_length', data = iris_df, hue = 'species', palette = 'plasma')
```

##### Pie Chart

```python
iris_df['species'].value_counts()
```

```python
plt.style.use("ggplot")
plt.figure(figsize=(9,9))
plt.pie(iris_df['species'].value_counts(), autopct='%0.2f%%', labels = ['setosa','versicolor','verginica'],
        explode = (0, 0.1, 0))
plt.show()
```

##### Distributional Charts

    * Density & Histogram Plots

```python
sns.set_style("whitegrid")
plt.figure(figsize=(12,9))
sns.distplot(boston_df['MEDV'], color = 'darkviolet')
plt.show()
```

##### Q-Q Plot

```python
import statsmodels.api as stats
stats.qqplot(boston_df['MEDV'] , line = '45', fit = True)
plt.show()
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(12,9))
sns.distplot(boston_df['AGE'], color = 'darkviolet')
plt.show()
```

##### Boxplot

```python
plt.style.use("ggplot")
plt.figure(figsize=(12,9))
sns.boxplot(data = iris_df)
```

```python
plt.style.use("ggplot")
plt.figure(figsize=(12,9))
sns.boxplot(data = iris_df, orient = 'h', palette = 'Reds')
```
