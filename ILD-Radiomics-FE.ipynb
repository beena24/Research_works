```python
import os
import pandas as pd
import cv2
import pydicom
import numpy as np

# Define the paths
images_path = 'D:/ILD_DB/data/cysts/images'
masks_path = 'D:/ILD_DB/data/cysts/mask'
segmented_path = 'D:/ILD_DB/data/cysts/segmented-cysts'
mapping_file = 'D:/ILD_DB/data/mapping.csv'

# Create the segmented directory if it doesn't exist
if not os.path.exists(segmented_path):
    os.makedirs(segmented_path)

# Read the mapping file
mapping_df = pd.read_csv(mapping_file)

# Function to read a DICOM file and return the pixel array with scaling
def read_dicom(dicom_path):
    try:
        dicom = pydicom.dcmread(dicom_path)
        img = dicom.pixel_array.astype(np.float32)
        
        # Apply rescale slope and intercept if present
        intercept = dicom.RescaleIntercept if 'RescaleIntercept' in dicom else 0
        slope = dicom.RescaleSlope if 'RescaleSlope' in dicom else 1
        img = slope * img + intercept
        
        return img
    except Exception as e:
        print(f"Error reading DICOM file {dicom_path}: {e}")
        return None

# Function to create segmented mask
def create_segmented_mask(image_path, mask_path, output_path):
    image = read_dicom(image_path)
    mask = read_dicom(mask_path)
    
    if image is None or mask is None:
        print(f"Skipping {image_path} or {mask_path} due to read error.")
        return
    
    if image.shape != mask.shape:
        print(f"Shape mismatch between image and mask: {image_path} and {mask_path}")
        return
    
    # Normalize the image for better visibility
    image = cv2.normalize(image, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
    
    # Apply the mask to the image
    segmented = cv2.bitwise_and(image, image, mask=mask.astype(np.uint8))
    
    # Save the segmented image as a PNG file
    success = cv2.imwrite(output_path, segmented)
    
    if success:
        print(f"Saved segmented image to {output_path}")
    else:
        print(f"Failed to save segmented image to {output_path}")

# Process each image and corresponding mask
for _, row in mapping_df.iterrows():
    image_name = row['image_name']
    mask_name = row['mask_name']
    
    image_path = os.path.join(images_path, image_name)
    mask_path = os.path.join(masks_path, mask_name)
    output_path = os.path.join(segmented_path, image_name.replace('.dcm', '.png'))

    # Debugging output
    if not os.path.exists(image_path):
        print(f'Image not found: {image_path}')
    if not os.path.exists(mask_path):
        print(f'Mask not found: {mask_path}')
    
    if os.path.exists(image_path) and os.path.exists(mask_path):
        create_segmented_mask(image_path, mask_path, output_path)
    else:
        if not os.path.exists(image_path):
            print(f'Image not found: {image_name}')
        if not os.path.exists(mask_path):
            print(f'Mask not found: {mask_name}')

print("Processing complete!")
```

```python
import os
import pandas as pd
import numpy as np
import pydicom
import SimpleITK as sitk
from radiomics import featureextractor
import matplotlib.pyplot as plt

# Define paths
mapping_file = 'D:/ILD_DB/ILD/fibrosis/mapping2.csv'
segmented_path = 'D:/ILD_DB/ILD/fibrosis/segmented_fibrosis'
output_file = 'D:/ILD_DB/ILD/fibrosis/features2.csv'

# Read the mapping file
mapping_df = pd.read_csv(mapping_file)

# Initialize the feature extractor with custom settings
extractor = featureextractor.RadiomicsFeatureExtractor()
extractor.settings['label'] = 255  # Update this to match your mask's label

# Function to read a DICOM file and return the pixel array as SimpleITK image
def read_dicom_to_sitk(dicom_path):
    try:
        dicom = pydicom.dcmread(dicom_path)
        array = dicom.pixel_array
        sitk_image = sitk.GetImageFromArray(array)
        sitk_image.SetSpacing([float(x) for x in dicom.PixelSpacing])  # Set spacing if available
        return sitk_image
    except Exception as e:
        print(f"Error reading DICOM file {dicom_path}: {e}")
        return None

# Function to show image
def show_image(image_path):
    sitk_image = sitk.ReadImage(image_path)
    array = sitk.GetArrayFromImage(sitk_image)
    plt.imshow(array[0], cmap='gray')
    plt.title(f"Image: {image_path}")
    plt.show()

# Function to check mask labels
def check_mask_labels(mask_path):
    sitk_image = sitk.ReadImage(mask_path)
    array = sitk.GetArrayFromImage(sitk_image)
    unique_labels = np.unique(array)
    print(f"Unique labels in mask {mask_path}: {unique_labels}")

# Function to extract features from a segmented mask
def extract_features(image_path, mask_path):
    image_sitk = read_dicom_to_sitk(image_path)
    mask_sitk = read_dicom_to_sitk(mask_path)
    
    if image_sitk is None or mask_sitk is None:
        print(f"Skipping {image_path} or {mask_path} due to read error.")
        return None
    
    try:
        result = extractor.execute(image_sitk, mask_sitk)
    except ValueError as e:
        print(f"Error extracting features from {image_path} and {mask_path}: {e}")
        return None
    
    return result

# List to hold extracted features
features_list = []

# Process each image and corresponding mask
for _, row in mapping_df.iterrows():
    image_name = row['image_name']
    mask_name = row['mask_name']
    
    image_path = os.path.join(segmented_path, image_name)
    mask_path = os.path.join(segmented_path, mask_name)
    
    if os.path.exists(image_path) and os.path.exists(mask_path):
        # Optional: Show and check images and masks
        # show_image(image_path)
        # show_image(mask_path)
        # check_mask_labels(mask_path)

        features = extract_features(image_path, mask_path)
        if features:
            features['image_name'] = image_name
            features['mask_name'] = mask_name
            features_list.append(features)
    else:
        if not os.path.exists(image_path):
            print(f'Image not found: {image_path}')
        if not os.path.exists(mask_path):
            print(f'Mask not found: {mask_path}')

# Convert features to DataFrame
features_df = pd.DataFrame(features_list)

# Save features to a CSV file
features_df.to_csv(output_file, index=False)

print("Feature extraction complete!")
```

```python
import os
import pandas as pd
import numpy as np
import pydicom
import SimpleITK as sitk
from radiomics import featureextractor
import matplotlib.pyplot as plt

# Define paths
mapping_file = 'D:/ILD_DB/ILD/fibrosis/mapping2.csv'
segmented_path = 'D:/ILD_DB/ILD/fibrosis/segmented_fibrosis'
output_file = 'D:/ILD_DB/ILD/fibrosis/features2.csv'

# Read the mapping file
mapping_df = pd.read_csv(mapping_file)

# Initialize the feature extractor with custom settings
extractor = featureextractor.RadiomicsFeatureExtractor()
extractor.settings['label'] = 255  # Update this to match your mask's label

# Function to read a DICOM file and return the pixel array as SimpleITK image
def read_dicom_to_sitk(dicom_path):
    try:
        dicom = pydicom.dcmread(dicom_path)
        array = dicom.pixel_array
        sitk_image = sitk.GetImageFromArray(array)
        if hasattr(dicom, 'PixelSpacing'):
            sitk_image.SetSpacing([float(x) for x in dicom.PixelSpacing])  # Set spacing if available
        return sitk_image
    except Exception as e:
        print(f"Error reading DICOM file {dicom_path}: {e}")
        return None

# Function to show image
def show_image(image_path):
    sitk_image = sitk.ReadImage(image_path)
    array = sitk.GetArrayFromImage(sitk_image)
    plt.imshow(array[0], cmap='gray')
    plt.title(f"Image: {image_path}")
    plt.show()

# Function to check mask labels
def check_mask_labels(mask_path):
    sitk_image = sitk.ReadImage(mask_path)
    array = sitk.GetArrayFromImage(sitk_image)
    unique_labels = np.unique(array)
    print(f"Unique labels in mask {mask_path}: {unique_labels}")
    return unique_labels

# Function to extract features from a segmented mask
def extract_features(image_path, mask_path):
    image_sitk = read_dicom_to_sitk(image_path)
    mask_sitk = read_dicom_to_sitk(mask_path)
    
    if image_sitk is None or mask_sitk is None:
        print(f"Skipping {image_path} or {mask_path} due to read error.")
        return None
    
    unique_labels = check_mask_labels(mask_path)
    if extractor.settings['label'] not in unique_labels:
        print(f"No expected label {extractor.settings['label']} found in mask {mask_path}. Skipping.")
        return None
    
    try:
        result = extractor.execute(image_sitk, mask_sitk)
    except ValueError as e:
        print(f"Error extracting features from {image_path} and {mask_path}: {e}")
        return None
    
    return result

# List to hold extracted features
features_list = []

# Process each image and corresponding mask
successful_extractions = 0
for _, row in mapping_df.iterrows():
    image_name = row['image_name']
    mask_name = row['mask_name']
    
    image_path = os.path.join(segmented_path, image_name)
    mask_path = os.path.join(segmented_path, mask_name)
    
    if os.path.exists(image_path) and os.path.exists(mask_path):
        # Optional: Show and check images and masks
        # show_image(image_path)
        # show_image(mask_path)
        # check_mask_labels(mask_path)

        features = extract_features(image_path, mask_path)
        if features:
            features['image_name'] = image_name
            features['mask_name'] = mask_name
            features_list.append(features)
            successful_extractions += 1
    else:
        if not os.path.exists(image_path):
            print(f'Image not found: {image_path}')
        if not os.path.exists(mask_path):
            print(f'Mask not found: {mask_path}')

# Convert features to DataFrame
features_df = pd.DataFrame(features_list)

# Save features to a CSV file
features_df.to_csv(output_file, index=False)

print(f"Feature extraction complete! Successfully extracted features from {successful_extractions} pairs of images and masks.")
```

```python
import os
import pandas as pd
import numpy as np
import pydicom
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.preprocessing import image

# Define paths
mapping_file = 'D:/ILD_DB/data/mapping.csv'
segmented_path = 'D:/ILD_DB/data/segmented-cysts'
output_file = 'D:/ILD_DB/data/eff_deep_learning.csv'

# Load pre-trained EfficientNetB0 model
base_model = EfficientNetB0(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
model = tf.keras.Model(inputs=base_model.input, outputs=base_model.output)

# Function to read a DICOM file and convert to RGB image
def read_dicom_as_rgb(dicom_path):
    try:
        dicom = pydicom.dcmread(dicom_path)
        img_array = dicom.pixel_array
        img_array = np.expand_dims(img_array, axis=-1)  # Add channel dimension if needed
        img_array = np.repeat(img_array, 3, axis=-1)  # Convert grayscale to RGB
        img_array = tf.image.resize(img_array, [224, 224])  # Resize to match model input
        return img_array.numpy()
    except Exception as e:
        print(f"Error reading DICOM file {dicom_path}: {e}")
        return None

# Function to preprocess and extract features from a CT image
def extract_features_ct_image(image_path):
    try:
        img_array = read_dicom_as_rgb(image_path)
        if img_array is None:
            return None
        img_array = preprocess_input(img_array)  # Preprocess the image
        img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension

        features = model.predict(img_array)
        features = features.flatten()  # Flatten the features for CSV
        return features
    except Exception as e:
        print(f"Error processing image {image_path}: {e}")
        return None

# Read the mapping file
mapping_df = pd.read_csv(mapping_file)

# List to hold extracted features
features_list = []

# Process each image and corresponding mask
for _, row in mapping_df.iterrows():
    image_name = row['image_name']
    mask_name = row['mask_name']  # Not used in feature extraction here, but could be used for advanced processing

    image_path = os.path.join(segmented_path, image_name)
    
    if os.path.exists(image_path):
        features = extract_features_ct_image(image_path)
        if features is not None:
            feature_dict = {'image_name': image_name}
            feature_dict.update({f'feature_{i}': features[i] for i in range(len(features))})
            features_list.append(feature_dict)
    else:
        print(f'Image not found: {image_path}')

# Convert features to DataFrame
features_df = pd.DataFrame(features_list)

# Save features to a CSV file
features_df.to_csv(output_file, index=False)

print("Feature extraction complete and saved to CSV!")
```

```python
import os
import pandas as pd
import numpy as np
import pydicom
import SimpleITK as sitk
from radiomics import featureextractor
import matplotlib.pyplot as plt

# Define paths
mapping_file = 'D:/ILD_DB/ILD/healthy/mapping4.csv'
segmented_path = 'D:/ILD_DB/ILD/healthy/segmented_healthy'
output_file = 'D:/ILD_DB/ILD/healthy/features4.csv'

# Read the mapping file
mapping_df = pd.read_csv(mapping_file)

# Initialize the feature extractor with custom settings
extractor = featureextractor.RadiomicsFeatureExtractor()
extractor.settings['label'] = 255  # Update this to match your mask's label

# Function to read a DICOM file and return the pixel array as SimpleITK image
def read_dicom_to_sitk(dicom_path):
    try:
        dicom = pydicom.dcmread(dicom_path)
        array = dicom.pixel_array
        sitk_image = sitk.GetImageFromArray(array)
        sitk_image.SetSpacing([float(x) for x in dicom.PixelSpacing])  # Set spacing if available
        return sitk_image
    except Exception as e:
        print(f"Error reading DICOM file {dicom_path}: {e}")
        return None

# Function to show image
def show_image(image_path):
    sitk_image = sitk.ReadImage(image_path)
    array = sitk.GetArrayFromImage(sitk_image)
    plt.imshow(array[0], cmap='gray')
    plt.title(f"Image: {image_path}")
    plt.show()

# Function to check mask labels
def check_mask_labels(mask_path):
    sitk_image = sitk.ReadImage(mask_path)
    array = sitk.GetArrayFromImage(sitk_image)
    unique_labels = np.unique(array)
    print(f"Unique labels in mask {mask_path}: {unique_labels}")

# Function to extract features from a segmented mask
def extract_features(image_path, mask_path):
    image_sitk = read_dicom_to_sitk(image_path)
    mask_sitk = read_dicom_to_sitk(mask_path)
    
    if image_sitk is None or mask_sitk is None:
        print(f"Skipping {image_path} or {mask_path} due to read error.")
        return None
    
    try:
        result = extractor.execute(image_sitk, mask_sitk)
    except ValueError as e:
        print(f"Error extracting features from {image_path} and {mask_path}: {e}")
        return None
    
    return result

# List to hold extracted features
features_list = []

# Process each image and corresponding mask
for _, row in mapping_df.iterrows():
    image_name = row['image_name']
    mask_name = row['mask_name']
    
    # Skip if image_name or mask_name is NaN or None
    if pd.isna(image_name) or pd.isna(mask_name):
        print(f"Skipping due to missing image or mask name: Image - {image_name}, Mask - {mask_name}")
        continue
    
    image_name = str(image_name)
    mask_name = str(mask_name)
    
    image_path = os.path.join(segmented_path, image_name)
    mask_path = os.path.join(segmented_path, mask_name)
    
    if os.path.exists(image_path) and os.path.exists(mask_path):
        # Optional: Show and check images and masks
        # show_image(image_path)
        # show_image(mask_path)
        # check_mask_labels(mask_path)

        features = extract_features(image_path, mask_path)
        if features:
            features['image_name'] = image_name
            features['mask_name'] = mask_name
            features_list.append(features)
    else:
        if not os.path.exists(image_path):
            print(f'Image not found: {image_path}')
        if not os.path.exists(mask_path):
            print(f'Mask not found: {mask_path}')

# Convert features to DataFrame
features_df = pd.DataFrame(features_list)

# Save features to a CSV file
features_df.to_csv(output_file, index=False)

print("Feature extraction complete!")

   
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import SMOTE

# Load the data
data = pd.read_csv('D:/ILD_DB/ILD/combined-features.csv')

# Inspect the class distribution
print(data['class-label'].value_counts())

# Separate features and labels
X = data.drop('class-label', axis=1)
y = data['class-label']

# Handle missing values if necessary
# X = X.fillna(X.mean())  # Example for filling missing values with mean

# Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Handle class imbalance using SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=42, stratify=y_resampled)

# Train a Random Forest classifier
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Evaluate the model
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))

# Optional: Tune hyperparameters using GridSearchCV
# from sklearn.model_selection import GridSearchCV
# param_grid = {
#     'n_estimators': [100, 200, 300],
#     'max_depth': [10, 20, 30],
#     'min_samples_split': [2, 5, 10]
# }
# grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=3, scoring='accuracy')
# grid_search.fit(X_train, y_train)
# print(grid_search.best_params_)
# best_model = grid_search.best_estimator_
# y_pred_best = best_model.predict(X_test)
# print(classification_report(y_test, y_pred_best))
# print(confusion_matrix(y_test, y_pred_best))
```
